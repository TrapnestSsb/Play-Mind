## Pregunta 1

**Comparación de algoritmos:** ¿qué modelo obtuvo el mejor `F1 Macro` y `AUC OvR` tanto en validación cruzada (5 particiones) como en prueba (20%)?

**Respuesta del equipo:**

De los cuatro algoritmos que probamos, **Gradient Boosting obtuvo los mejores resultados**. En la validación cruzada de cinco particiones consiguió un **F1 Macro de 0.9075** y un **AUC OvR de 0.9440**. En la prueba con el 20 % de los datos obtuvo **F1 Macro de 0.9051**, **AUC OvR de 0.9429** y una exactitud de **0.9101** (91.01 %).

Como equipo decidimos usar **Gradient Boosting** para PlayMind, porque fue el que mejor clasificó los tres niveles de participación en nuestros resultados. Aprendimos que comparar varios algoritmos es importante, ya que no todos funcionan igual con los mismos datos.


## Pregunta 2

**Análisis de sobreajuste:** ¿cuál fue la brecha entre `Train F1` y `Test F1` en cada algoritmo (especialmente en Random Forest vs. Gradient Boosting)?

**Respuesta del equipo:**

Comparamos el F1 Macro de entrenamiento con el de prueba para revisar si algún algoritmo se había ajustado demasiado a los datos con los que aprendió. Estas fueron las diferencias:

| Algoritmo | Train F1 | Test F1 | Brecha |
|---|---:|---:|---:|
| Regresión Logística | 0.8187 | 0.8111 | 0.0076 |
| Random Forest | 0.9239 | 0.8812 | 0.0428 |
| Gradient Boosting | 0.9201 | 0.9051 | 0.0150 |
| KNN (k=15) | 1.0000 | 0.7783 | 0.2217 |

El caso más llamativo fue **KNN**, porque obtuvo 1.0000 en entrenamiento, pero bajó bastante en prueba. Entre **Random Forest y Gradient Boosting**, Random Forest tuvo una brecha mayor (**0.0428 frente a 0.0150**). Por eso, Gradient Boosting nos pareció más confiable para clasificar jugadores nuevos, aunque tendremos que seguir probándolo.


## Pregunta 3

**Matriz de confusión y errores:** ¿qué niveles se confunden con mayor frecuencia (por ejemplo, `Medium` con `Low` o `High`) y qué implicación tiene clasificar erróneamente a un jugador `Low`?

**Respuesta del equipo:**

Con el modelo elegido, **Gradient Boosting**, hubo **720 errores entre 8,007 jugadores de prueba (8.99 %)**. Las confusiones más comunes fueron clasificar a **Low como Medium (185 casos)** y a **High como Medium (178 casos)**. También encontramos **119 casos de Medium como High** y **102 de Medium como Low**.

Lo que más nos preocupa para PlayMind es que **249 jugadores que realmente eran Low fueron clasificados como Medium o High** (185 + 64). Si eso ocurriera en nuestro prototipo, podrían no recibir las misiones cortas o ayudas pensadas para quienes participan menos. Aprendimos que una buena exactitud general no significa que el sistema nunca se equivoque con los jugadores que más nos interesa atender.


## Pregunta 4

**Conductas que anticipan el abandono (`Low`):** según la importancia por permutación y los promedios por clase, ¿qué variables pesan más (sesiones por semana, duración media de sesión, nivel o compras)?

**Respuesta del equipo:**

Las variables más importantes fueron **SessionsPerWeek (sesiones por semana)**, con una caída de **0.4351** en F1 Macro al desordenarla, y **AvgSessionDurationMinutes (duración promedio de sesión)**, con **0.3584**. Después quedaron bastante más abajo **PlayerLevel (0.0258)** y **AchievementsUnlocked (0.0253)**; **InGamePurchases (compras)** tuvo una importancia cercana a cero.

También vimos que los jugadores **Low** promediaron **4.53 sesiones por semana y 66.88 minutos por sesión**, mientras que los **High** tuvieron **14.25 sesiones y 131.92 minutos**. Esto nos hizo pensar que la frecuencia y duración de las partidas ayudan mucho a reconocer la participación baja. Sin embargo, **no podemos asegurar que un jugador abandonará el juego**, porque el dataset mide participación y no abandono real.


## Pregunta 5

**Cambio de diseño propuesto:** ¿qué ajuste concreto en el videojuego (dificultad adaptativa, misiones cortas o recompensas tempranas) se probaría para retener a los jugadores detectados en nivel `Low`?

**Respuesta del equipo:**

Como equipo proponemos probar **misiones cortas de 5 a 10 minutos y recompensas tempranas** para los jugadores que el modelo clasifique como **Low**. También podríamos ofrecer un tutorial opcional y bajar un poco la dificultad si el jugador falla varias veces.

Elegimos estas ideas porque observamos que el grupo Low juega menos sesiones y durante menos tiempo. Más adelante revisaríamos si, después de estos cambios, aumentan sus sesiones semanales o la duración de sus partidas. Por ahora son **propuestas por probar**, no mejoras que ya hayamos comprobado.
